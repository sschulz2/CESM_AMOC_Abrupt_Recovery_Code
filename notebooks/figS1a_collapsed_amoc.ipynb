# Figure S1a: Eulerian-mean Atlantic streamfunction during the collapse, 8xCO2

Mean over years 800-2299 since 8xCO2.

Input: `AMOC_mean_800-2299_CESM104_abrupt8x_5100.nc`

```python
import numpy as np
import xarray as xr
import matplotlib
import matplotlib.pyplot as plt
import common
import cmocean as cm

matplotlib.rc('font', size=16)
```

```python
moc = xr.open_dataset(common.data_path('AMOC_mean_800-2299_CESM104_abrupt8x_5100.nc')).MOC
depth = -moc.moc_z / 100  # moc_z is in cm, positive down
```

```python
color_limit = 20  # Sv
plt.subplots(figsize=(13, 6))
p = plt.contourf(moc.lat_aux_grid, depth, moc, vmax=color_limit, vmin=-color_limit, cmap=cm.cm.balance)
plt.contour(moc.lat_aux_grid, depth, moc, vmax=color_limit, vmin=-color_limit, colors='black')
plt.colorbar(p, label='MOC Strength [Sv]')
plt.ylim([-2000, 0])
plt.xlim([-35, 60])
plt.ylabel('z [m]')
plt.xlabel('Latitude')
plt.savefig(common.figure_path('figS1a_collapsed_amoc.png'), dpi=150, bbox_inches='tight')
plt.show()
```
